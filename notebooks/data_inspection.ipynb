**Error analysis in Datasets**

```python
import pandas as pd
import numpy as np
import os

# 1. Setup paths for the processed data
processed_path = 'data/processed'
if not os.path.exists(processed_path):
    os.makedirs(processed_path)

print("--- SECTION 2: STARTING DATA CLEANING & PREPARATION ---\n")

# 2. Standardize Categories & Remove Duplicates (Global Function)
def initial_clean(df):
    # Remove exact duplicate rows
    df = df.drop_duplicates()
    # Standardize text: lowercase and strip whitespace
    for col in df.select_dtypes(include=['object']).columns:
        df[col] = df[col].astype(str).str.lower().str.strip()
    return df

# Apply initial cleaning to all loaded dataframes
app_events = initial_clean(app_events)
complaints = initial_clean(complaints)
customers = initial_clean(customers)
deliveries = initial_clean(deliveries)
drivers = initial_clean(drivers)
hubs = initial_clean(hubs)
incidents = initial_clean(incidents)
orders = initial_clean(orders)
vehicles = initial_clean(vehicles)

# 3. Fix Missing Values (Imputation)
# Battery Health: Fill with median to avoid outliers
vehicles['battery_health_pct'] = pd.to_numeric(vehicles['battery_health_pct'], errors='coerce')
vehicles['battery_health_pct'] = vehicles['battery_health_pct'].fillna(vehicles['battery_health_pct'].median())

# Customer Rating: Fill with mean
deliveries['customer_rating_post_delivery'] = pd.to_numeric(deliveries['customer_rating_post_delivery'], errors='coerce')
deliveries['customer_rating_post_delivery'] = deliveries['customer_rating_post_delivery'].fillna(deliveries['customer_rating_post_delivery'].mean())

# 4. Convert Dates/Times
# This is crucial for calculating delays and operational patterns
deliveries['dispatch_time'] = pd.to_datetime(deliveries['dispatch_time'], errors='coerce')
deliveries['delivery_completed_at'] = pd.to_datetime(deliveries['delivery_completed_at'], errors='coerce')
orders['order_created_at'] = pd.to_datetime(orders['order_created_at'], errors='coerce')

# 5. Create Engineered Features (The "Analytical" part)
# Feature A: Delivery Duration in Minutes
deliveries['delivery_duration_mins'] = (deliveries['delivery_completed_at'] - deliveries['dispatch_time']).dt.total_seconds() / 60

# Feature B: Delay Flag (If duration is longer than the promised window in orders)
#  merge briefly to calculate this
order_windows = orders[['order_id', 'promised_window_hours']].copy()
deliveries = deliveries.merge(order_windows, on='order_id', how='left')
deliveries['proof_of_completion_missing'] = np.where(deliveries['delivery_duration_mins'] > (deliveries['promised_window_hours'] * 60), 1, 0)

# Feature C: Troubleshooting App Events (Count events per order/customer)
event_counts = app_events.groupby('customer_id').size().reset_index(name='total_app_interactions')
customers = customers.merge(event_counts, on='customer_id', how='left').fillna(0)

# 6. Save Cleaned Datasets
cleaned_files = {
    'app_events': app_events, 'complaints': complaints, 'customers': customers,
    'deliveries': deliveries, 'drivers': drivers, 'hubs': hubs,
    'incidents': incidents, 'orders': orders, 'vehicles': vehicles
}

print("Exporting cleaned data to /data/processed...")
for name, df in cleaned_files.items():
    path = os.path.join(processed_path, f"{name}_cleaned.csv")
    df.to_csv(path, index=False)
    print(f" - {name}_cleaned.csv saved.")

print("\n--- DATA CLEANING COMPLETE ---")
```
